This file is a simple test to ensure that the random_state used to create the modeling data was implemented correctly, and rerunning the script yields the same result.

```python
import pandas as pd

real_df = pd.read_parquet('data/modeling/modeling_train.parquet')
test_df = pd.read_parquet('data/modeling_check/modeling_train.parquet')
```

```python
print(f'Real shape: {real_df.shape}')
print(f'Test shape: {test_df.shape}')

real_df.shape == test_df.shape
```

```python
sorted_real = real_df.sort_values(by = ['serial_number', 'window'], ignore_index=True, ascending = True)
sorted_test = test_df.sort_values(by = ['serial_number', 'window'], ignore_index=True, ascending = True)
```

```python
sorted_real.equals(sorted_test)
```
